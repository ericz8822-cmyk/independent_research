Here we want to visualize the 1,000,000 prime number from first_1000000_primes.csv

```python
# importing the data as a list of 1000000 tuples of two numbers
import csv

with open("first_1000000_primes.csv", mode="r", newline="") as file:
  reader = csv.reader(file)
  next(reader)  # Skips the header row
  data = [(int(row[0]), int(row[1])) for row in reader]
print(data[:10])
```
