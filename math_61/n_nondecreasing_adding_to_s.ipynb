```python
import math
%pip install pandas

def f(n, s):
    if not (isinstance(n, int) & isinstance(s, int)):
        print("Both inputs should be integers.")
        return
    if n < 1:
        print("n must be at least 1.")
        return
    if s == 0 or n == 1:
        return 1
    max = math.floor(s/n)
    sum = 0
    if max == 0:
        return f(s,s)
    for i in range(max):
        sum += f(n-1, s-i*n)
    return sum
print(f(10,10))
```

```python
import pandas as pd
def print_list_explicit(size_list):
    list = []
    for i in range(size_list):
        list.append([])
        for j in range(size_list):
            list[i].append((f(i+1, j)))
    df = pd.DataFrame(list)
    df.index = [f'{i+1}' for i in range(size_list)]
    return df
def print_list_truncated(size_list):
    list = []
    for i in range(size_list):
        list.append([])
        for j in range(size_list):
            if j >= 2 and i >= (j-1)/2:
                list[i].append('-')
            else:
                list[i].append((f(i+1, j)))
    df = pd.DataFrame(list)
    df.index = [f'{i+1}' for i in range(size_list)]
    df = df.iloc[:, 2:]
    return df
```
